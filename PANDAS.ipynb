```python
import pandas as pd
```

```python
# Sample Data
data = {'Category': ['A', 'A', 'B', 'B', 'C'],
        'Sales': [100, 200, 150, 300, 250]}

df = pd.DataFrame(data)

# Performing single aggregations
print(df['Sales'].sum())    # Total sum of Sales
print(df['Sales'].mean())   # Average Sales
print(df['Sales'].max())    # Maximum Sales
print(df['Sales'].min())    # Minimum Sales
print(df['Sales'].count())  # Count of Sales values
print(df['Sales'].median()) # Median of Sales
print(df['Sales'].std())    # Standard Deviation of Sales
print(df['Sales'].var())    # variance of sales
```

```python
import pandas as pd

# Sample Data
data = {'Category': ['A', 'A', 'B', 'B', 'C'],
        'Sales': [100, 200, 150, 300, 250]}

df = pd.DataFrame(data)

# Grouping by Category and calculating the total Sales
grouped = df.groupby('Category')['Sales'].sum()
print(grouped)
```

```python
data = {'Category': ['A', 'A', 'B', 'B', 'C', 'C'],
        'Region': ['East', 'West', 'East', 'West', 'East', 'West'],
        'Sales': [100, 200, 150, 300, 250, 350]}

df = pd.DataFrame(data)

# Grouping by Category and Region, calculating sum of Sales
grouped = df.groupby(['Category', 'Region'])['Sales'].sum()
print(grouped)
```

```python
# Group by Category and apply multiple aggregation functions
grouped = df.groupby('Category').agg({'Sales': ['sum', 'mean', 'max']})
print(grouped)
```

```python
import pandas as pd

# Sample Data
data = {'Category': ['A', 'A', 'B', 'B', 'C'],
        'Sales': [100, 200, 150, 300, 250],
        'Profit': [10, 20, 15, 30, 25]}

df = pd.DataFrame(data)

# Filter where Sales > 150 AND Profit > 15
filtered_df = df[(df['Sales'] > 150) & (df['Profit'] > 15)]
print(filtered_df)
```

```python
filtered_df = df.query("Sales > 150 & Profit > 15")
print(filtered_df)
```

```python
filtered_df = df[df['Category'].isin(['A', 'C'])]  # Selects rows where Category is 'A' or 'C'
print(filtered_df)
```

```python
# First Table (df1)
data1 = {'ID': [1, 2, 3, 4],
         'Name': ['Alice', 'Bob', 'Charlie', 'David'],
         'Sales': [5000, 7000, 6000, 8000]}
df1 = pd.DataFrame(data1)

# Second Table (df2)
data2 = {'ID': [3, 4, 5, 6],
         'City': ['New York', 'Los Angeles', 'Chicago', 'Houston'],
         'Profit': [400, 500, 300, 700]}
df2 = pd.DataFrame(data2)

# **INNER JOIN** (Keeps only matching rows)
inner_join = pd.merge(df1, df2, on="ID", how="inner")
print(inner_join)
```

```python
left_join = pd.merge(df1, df2, on="ID", how="left")
print(left_join)
```

```python
right_join = pd.merge(df1, df2, on="ID", how="right")
print(right_join)
```

```python
outer_join = pd.merge(df1, df2, on="ID", how="outer")
print(outer_join)
```

```python
df1 = pd.DataFrame({'Emp_ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df2 = pd.DataFrame({'Employee_ID': [2, 3, 4], 'Salary': [50000, 60000, 70000]})

merged_df = pd.merge(df1, df2, left_on="Emp_ID", right_on="Employee_ID", how="inner")
print(merged_df)
```
